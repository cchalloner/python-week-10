### Program B

**Intended behavior:** Return the average of a list of numbers. If the list is empty, print an error message and return `None`.

Your fixed version should print `20.0`, then the error message, then `None`.


````python
# Program B: BUGGY - fix it
def average(numbers):
    try:
        return sum(numbers) / len(numbers)
    except ValueError:
        print("Error: Cannot average an empty list.")
        return None


print(average([10, 20, 30]))
print(average([]))
````
